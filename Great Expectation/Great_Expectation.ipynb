```python
# Create a data context

from great_expectations.data_context import FileDataContext

context = FileDataContext.create(project_root_dir='./')
```

```python
datasource_name = 'amazon_datasource'
datasource = context.sources.add_pandas(datasource_name)

asset_name = 'amazon_ecommerce_data'
path_to_data = '/content/opt/airflow/data/amazon_ecommerce_1M.csv'

asset = datasource.add_csv_asset(
    name=asset_name,
    filepath_or_buffer=path_to_data
)

batch_request = asset.build_batch_request()
```

```python
# Create expectation suite
expectation_suite_name = 'amazon_suite'
context.add_or_update_expectation_suite(expectation_suite_name)

# Create validator
validator = context.get_validator(batch_request=batch_request, expectation_suite_name=expectation_suite_name)

# Check data
validator.head()
```

```python
validator.expect_column_values_to_be_unique("user_id") ## 1. checking user_id is unique
```

```python
validator.expect_column_values_to_be_between('rating', min_value=1, max_value=5) #2 Rating must be range 1 - 5
```

```python
validator.expect_column_values_to_be_in_set("payment_method", ["UPI", "Credit Card", "Debit Card", "Cash on Delivery"]) ##3. Check payment method must be UPI,credit card, debit card, and Cash on delivery
```

```python
validator.expect_column_values_to_be_in_type_list('stock', ['int']) ##4. stock must be integer
```

```python
validator.expect_column_values_to_not_be_in_set("price", [0]) ##5 Price cannot be 0
```

```python
validator.expect_column_value_lengths_to_be_between('brand', min_value=2) #6  ##Brand minimum Character Must be 2 Character
```

```python
validator.expect_column_values_to_match_strftime_format("purchase_date", "%Y-%m-%d") ##Purchase date must be YYYY-MM-DD
```

```python
df.to_csv("P2M3_Mardi_Wicaksana_data_raw.csv") ##Save data to csv
```

```python
# Create a checkpoint

checkpoint_1 = context.add_or_update_checkpoint(
    name = 'checkpoint_1',
    validator = validator,
)
```

```python
# Run a checkpoint

checkpoint_result = checkpoint_1.run()
```
